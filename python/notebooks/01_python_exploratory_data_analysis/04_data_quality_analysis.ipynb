### Import libraries:

```python
import pandas as pd
import numpy as np
```

### Load all five datasets:

```python
df_feature_adoption = pd.read_csv("../data/feature_adoption.csv")

df_user_engagement = pd.read_csv("../data/user_engagement.csv")

df_churn_analysis = pd.read_csv("../data/churn_analysis.csv")

df_api_performance = pd.read_csv("../data/api_performance.csv")

df_experiment_analysis = pd.read_csv("../data/experiment_analysis.csv")
```

### Create a dataset dictionary:

```python
datasets = {
    "Feature Adoption": df_feature_adoption,
    "User Engagement": df_user_engagement,
    "Churn Analysis": df_churn_analysis,
    "API Performance": df_api_performance,
    "Experiment Analysis": df_experiment_analysis
}
```

### Check missing values:

```python
for name, df in datasets.items():

    print(f"\n{name}")
    print(df.isnull().sum())
    print("-" * 40)
```

### Calculate missing value percentages:

```python
for name, df in datasets.items():

    missing_percentage = (
        df.isnull().sum() / len(df)
    ) * 100

    print(f"\n{name} - Missing Value Percentage")
    print(missing_percentage.round(2))
    print("-" * 40)
```

### Create a Missing Values Summary:

```python
missing_summary = pd.DataFrame()

for name, df in datasets.items():

    temp = pd.DataFrame({
        "Dataset": name,
        "Column": df.columns,
        "Missing Values": df.isnull().sum().values,
        "Missing Percentage": (
            df.isnull().sum().values / len(df)
        ) * 100
    })

    missing_summary = pd.concat(
        [missing_summary, temp],
        ignore_index=True
    )

missing_summary = missing_summary[
    missing_summary["Missing Values"] > 0
]

missing_summary
```

### Check Duplicate Records:

```python
for name, df in datasets.items():

    print(f"\n{name}")
    print("Duplicate Rows:", df.duplicated().sum())
    print("-" * 40)
```

### Check Duplicate User IDs:

```python
print(
    "Duplicate User IDs:",
    df_user_engagement["user_id"].duplicated().sum()
)
```

```python
print(
    "Missing User IDs:",
    df_user_engagement["user_id"].isnull().sum()
)
```

### Check Invalid Numeric Values:

#### inspect numerical columns:

```python
df_user_engagement.select_dtypes(
    include="number"
).describe()
```

#### Check for negative values in selected count columns:

```python
count_columns = [
    "active_days",
    "total_events",
    "total_sessions",
    "total_api_requests",
    "successful_requests",
    "failed_requests"
]

for column in count_columns:

    if column in df_user_engagement.columns:

        print(
            column,
            "Negative values:",
            (df_user_engagement[column] < 0).sum()
        )
```

### Check API Performance Values:

```python
print(
    "Error rates below 0:",
    (df_api_performance["error_rate_percentage"] < 0).sum()
)

print(
    "Error rates above 100:",
    (df_api_performance["error_rate_percentage"] > 100).sum()
)
```

#### Check response times:

```python
print(
    "Negative response times:",
    (df_api_performance["average_response_time_ms"] < 0).sum()
)
```

### Check Experiment Outcome Values:

```python
binary_columns = [
    "sdk_integrated",
    "first_api_request",
    "adopted_feature",
    "retained_d7"
]

for column in binary_columns:

    if column in df_experiment_analysis.columns:

        print(f"\n{column}")
        print(df_experiment_analysis[column].value_counts(dropna=False))
```

### Create a Data Quality Report:

```python
quality_report = pd.DataFrame({
    "Dataset": list(datasets.keys()),

    "Total Rows": [
        df.shape[0] for df in datasets.values()
    ],

    "Total Columns": [
        df.shape[1] for df in datasets.values()
    ],

    "Missing Values": [
        df.isnull().sum().sum()
        for df in datasets.values()
    ],

    "Duplicate Rows": [
        df.duplicated().sum()
        for df in datasets.values()
    ]
})

quality_report
```

### Save the Data Quality Report:

```python
quality_report.to_csv(
    "../data/data_quality_report.csv",
    index=False
)

missing_summary.to_csv(
    "../data/missing_values_report.csv",
    index=False
)
```
