### Import libraries:

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from google.cloud import bigquery

print("All libraries imported successfully!")
```

### Connect to BigQuery:

```python
PROJECT_ID = "ecommerce-sales-nabin-2026"

client = bigquery.Client(
    project=PROJECT_ID
)

print("BigQuery connection successful!")
print("Project ID:", client.project)
```

### Verify your analytics dataset:

```python
DATASET_ID = "devflow_analytics"

dataset = client.get_dataset(
    f"{PROJECT_ID}.{DATASET_ID}"
)

print("Dataset:", dataset.dataset_id)
print("Location:", dataset.location)
```

### List all mart tables:

```python
tables = client.list_tables(
    f"{PROJECT_ID}.{DATASET_ID}"
)

for table in tables:
    print(table.table_id)
```
