### Import library and bigquery connection

```python
from google.cloud import bigquery
import pandas as pd
import numpy as np

PROJECT_ID = "ecommerce-sales-nabin-2026"

client = bigquery.Client(project=PROJECT_ID)
```

### 1. Load Feature Adoption Dataset:

```python
query = """
SELECT *
FROM `ecommerce-sales-nabin-2026.devflow_analytics.mart_feature_adoption`
"""

df_feature_adoption = client.query(query).to_dataframe()

print("Feature Adoption Dataset Loaded Successfully!")
print("Shape:", df_feature_adoption.shape)
```

### 2. Load User Engagement Dataset

```python
query = """
SELECT *
FROM `ecommerce-sales-nabin-2026.devflow_analytics.mart_user_engagement`
"""

df_user_engagement = client.query(query).to_dataframe()

print("User Engagement Dataset Loaded Successfully!")
print("Shape:", df_user_engagement.shape)
```

### 3. Load Churn Analysis Dataset

```python
query = """
SELECT *
FROM `ecommerce-sales-nabin-2026.devflow_analytics.mart_churn_analysis`
"""

df_churn_analysis = client.query(query).to_dataframe()

print("Churn Analysis Dataset Loaded Successfully!")
print("Shape:", df_churn_analysis.shape)
```

### 4. Load API Performance Dataset

```python
query = """
SELECT *
FROM `ecommerce-sales-nabin-2026.devflow_analytics.mart_api_performance`
"""

df_api_performance = client.query(query).to_dataframe()

print("API Performance Dataset Loaded Successfully!")
print("Shape:", df_api_performance.shape)
```

### 5. Load Experiment Analysis Dataset

```python
query = """
SELECT *
FROM `ecommerce-sales-nabin-2026.devflow_analytics.mart_experiment_analysis`
"""

df_experiment_analysis = client.query(query).to_dataframe()

print("Experiment Analysis Dataset Loaded Successfully!")
print("Shape:", df_experiment_analysis.shape)
```

### 6. Verify All Five Datasets

```python
datasets = {
    "Feature Adoption": df_feature_adoption,
    "User Engagement": df_user_engagement,
    "Churn Analysis": df_churn_analysis,
    "API Performance": df_api_performance,
    "Experiment Analysis": df_experiment_analysis
}

for name, df in datasets.items():

    print(f"\n{name}")
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("-" * 40)
```
