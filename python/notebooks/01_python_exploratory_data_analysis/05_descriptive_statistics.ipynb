### Import libraries:

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

### Load all five datasets:

```python
df_feature_adoption = pd.read_csv("../data/feature_adoption.csv")

df_user_engagement = pd.read_csv("../data/user_engagement.csv")

df_churn_analysis = pd.read_csv("../data/churn_analysis.csv")

df_api_performance = pd.read_csv("../data/api_performance.csv")

df_experiment_analysis = pd.read_csv("../data/experiment_analysis.csv")
```

### Create a dataset dictionary:

```python
datasets = {
    "Feature Adoption": df_feature_adoption,
    "User Engagement": df_user_engagement,
    "Churn Analysis": df_churn_analysis,
    "API Performance": df_api_performance,
    "Experiment Analysis": df_experiment_analysis
}
```

### Calculate Mean:

```python
numeric_df = df_user_engagement.select_dtypes(
    include="number"
)

mean_values = numeric_df.mean()

print("Mean Values:")
print(mean_values)
```

### Calculate Median:

```python
median_values = numeric_df.median()

print("Median Values:")
print(median_values)
```

### Calculate Mode:

```python
mode_values = numeric_df.mode()

print("Mode Values:")
print(mode_values)
```

#### For categorical engagement segments:

```python
print(
    df_user_engagement["engagement_segment"].mode()
)
```

### Calculate Standard Deviation:

```python
std_values = numeric_df.std()

print("Standard Deviation:")
print(std_values)
```

### Calculate Variance:

```python
variance_values = numeric_df.var()

print("Variance:")
print(variance_values)
```

### Calculate Minimum and Maximum:

```python
print("Minimum Values:")
print(numeric_df.min())

print("\nMaximum Values:")
print(numeric_df.max())
```

### Calculate Quartiles and Percentiles:

```python
quartiles = numeric_df.quantile(
    [0.25, 0.50, 0.75]
)

print(quartiles)
```

#### Calculate the 90th and 95th percentiles:

```python
percentiles = numeric_df.quantile(
    [0.90, 0.95, 0.99]
)

print(percentiles)
```

### Analyze User Engagement Segments:

#### calculate the number of users in each segment.

```python
segment_counts = df_user_engagement[
    "engagement_segment"
].value_counts()

print(segment_counts)
```

#### Calculate the percentage distribution

```python
segment_percentage = (
    df_user_engagement["engagement_segment"]
    .value_counts(normalize=True) * 100
)

print(segment_percentage.round(2))
```

### Analyze API Request Distribution:

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df_user_engagement,
    x="total_api_requests",
    bins=30,
    kde=True
)

plt.title("Distribution of Total API Requests")
plt.xlabel("Total API Requests")
plt.ylabel("Number of Users")

plt.tight_layout()
plt.show()
```

### Analyze Active Days Distribution:

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df_user_engagement,
    x="active_days",
    bins=30,
    kde=True
)

plt.title("Distribution of User Active Days")
plt.xlabel("Active Days")
plt.ylabel("Number of Users")

plt.tight_layout()
plt.show()
```
