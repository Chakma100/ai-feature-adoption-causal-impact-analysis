### Import required libraries:

```python
import pandas as pd
import numpy as np
```

### Load all five datasets:

```python
df_feature_adoption = pd.read_csv("../data/feature_adoption.csv")

df_user_engagement = pd.read_csv("../data/user_engagement.csv")

df_churn_analysis = pd.read_csv("../data/churn_analysis.csv")

df_api_performance = pd.read_csv("../data/api_performance.csv")

df_experiment_analysis = pd.read_csv("../data/experiment_analysis.csv")
```

### Create a dataset dictionary:

```python
datasets = {
    "Feature Adoption": df_feature_adoption,
    "User Engagement": df_user_engagement,
    "Churn Analysis": df_churn_analysis,
    "API Performance": df_api_performance,
    "Experiment Analysis": df_experiment_analysis
}
```

### Check Dataset Shape:

```python
for name, df in datasets.items():

    print(f"\n{name}")
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("Total Cells:", df.size)
    print("-" * 40)
```

### Check Column Names:

```python
for name, df in datasets.items():

    print(f"\n{name}")
    print("Columns:")
    print(df.columns.tolist())
    print("-" * 40)
```

### Check Data Types:

```python
for name, df in datasets.items():

    print(f"\n{name}")
    print(df.dtypes)
    print("-" * 40)
```

### Check Dataset Information:

```python
for name, df in datasets.items():

    print(f"\n{name}")
    df.info()
    print("-" * 50)
```

### Generate Descriptive Summary:

```python
for name, df in datasets.items():

    print(f"\n{name}")
    print(df.describe(include="all"))
    print("-" * 50)
```

### Create an Overall Dataset Summary:

```python
overview = pd.DataFrame({
    "Dataset": list(datasets.keys()),

    "Rows": [
        df.shape[0]
        for df in datasets.values()
    ],

    "Columns": [
        df.shape[1]
        for df in datasets.values()
    ],

    "Missing Values": [
        df.isnull().sum().sum()
        for df in datasets.values()
    ],

    "Duplicate Rows": [
        df.duplicated().sum()
        for df in datasets.values()
    ],

    "Memory Usage (KB)": [
        round(df.memory_usage(deep=True).sum() / 1024, 2)
        for df in datasets.values()
    ]
})

overview
```

### Save the Overview Report:

```python
overview.to_csv(
    "../data/dataset_overview_report.csv",
    index=False
)
```
